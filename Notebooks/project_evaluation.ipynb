# PROBLEM STATEMENT
Audience engagement with short-form Instagram Reels varies widely across videos, even within the same creator’s account. Because creators differ in audience size and baseline performance, raw engagement metrics alone do not provide a consistent basis for comparison. 

The purpose of this analysis is to identify which content characteristics are associated with a Reel achieving above-baseline performance, as defined by the binary variable `performance_bucket`. This creator-relative measure allows performance to be evaluated within the context of each creator’s typical output. 

By examining relationships between content features (such as structure, hook type, and presentation elements) and performance outcomes, this analysis aims to identify patterns associated with high-performing Reels that can inform data-driven content strategies.

---

# PREVIOUS WORK
The dataset was cleaned and prepared to ensure consistency, accuracy, and suitability for analysis. Initial checks confirmed that there were no significant missing or duplicate values, so no records were removed on that basis. Unnecessary columns, including hook_text and url, were dropped as they were not relevant to the analysis. The creator_handle variable was retained as an identifier to preserve the structure of the dataset, although it was not used directly in the analysis since it does not represent a measurable feature. Column names were standardized for clarity, and categorical values (e.g., craft categories and reel structures) were cleaned by converting to lowercase, removing extra spaces, and standardizing formatting to ensure consistency across categories.

Several variables originally stored as strings (e.g., views, likes, comments, and total_followers) were converted into numeric format to enable quantitative analysis. A new variable, engagement_rate, was created to measure engagement relative to views, providing a normalized metric for comparing performance across reels. Categorical variables were transformed into binary (dummy) variables to support analysis of specific features such as hook type, craft category, and reel structure. Additionally, binary indicators (e.g., pov_present, result_shown_first) were encoded as 0/1 values for consistency.

No outliers were removed, as extreme values in engagement metrics are meaningful in the context of social media performance. Finally, the target variable performance_bucket was converted into a binary format (0 = baseline, 1 = high-performing) to support comparative and predictive analysis.

---

# MODELS
Three classification models were attempted: logistic regression, a classification tree, and k-nearest neighbors (k-NN). These models were selected because the target variable was binary and the predictors were primarily content features encoded as binary indicators. Among the three, k-NN performed best overall, logistic regression provided a useful comparison because of its interpretability, and the classification tree performed worst.

The best model was k-nearest neighbors (k-NN). This conclusion was based on its stronger overall predictive performance compared with the other models, including higher accuracy and precision than logistic regression, while the classification tree showed the weakest results. Although logistic regression achieved a slightly better balance between precision and recall for the high-performing class, k-NN provided the strongest overall model because it classified reels more accurately and handled the distinction between baseline and high-performing reels more effectively. The results also suggested that content features alone had only modest predictive power, meaning that external factors such as timing, audience behavior, and algorithmic reach likely also influenced Reel performance.

One lesson from the business/research and data understanding phase is that dataset scope strongly affects model performance. ## Modeling and Evaluation

The newer tuned logistic regression model achieved the highest F1 score for the high-performing class (1) at 0.36, along with the highest recall for that class at 0.50. Because the goal of this analysis is to identify high-performing reels rather than simply maximize overall accuracy, logistic regression provides the strongest final model. It offered the best balance between precision and recall for the high-performing class and benefited the most from the addition of interaction features, suggesting that combinations of content characteristics matter more than isolated features alone.

However, the new results still suggest that content features have only modest predictive power, even after interaction terms were added. While the updated logistic regression model improved somewhat, all three models still struggled to consistently distinguish high-performing reels from baseline reels.
